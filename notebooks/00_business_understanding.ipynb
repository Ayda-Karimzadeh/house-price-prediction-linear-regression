# Business Understanding

This project aims to develop a machine learning model to predict housing prices using the California Housing Prices dataset, sourced from the 1990 California census.

**1. Business Problem**

Estimating housing prices accurately is a common and valuable capability for real-estate platforms, supporting more informed pricing and negotiation decisions for buyers and sellers.

This project uses a historical dataset (California housing data from the 1990 census) to build and validate a complete regression workflow — from data understanding through model evaluation — for predicting median house value from district-level demographic and housing features. Because the underlying data is over 30 years old, the resulting model is not intended to produce real-time, market-accurate price estimates. Instead, the project serves as a methodological foundation: the same pipeline (feature engineering, model selection, evaluation) is designed to be reapplied later to current, live housing data as part of a broader real-estate pricing platform.

**2. Project Objective**

The objective of this project is to build a regression model that predicts the median house value of a California housing district based on available demographic and housing-related features, and to demonstrate a rigorous, end-to-end ML workflow that can later be adapted to real, current data.

**3. Machine Learning Task**

* **Task:** Supervised Learning
* **Problem Type:** Regression
* **Target Variable:** `median_house_value`

**4. Features**

The dataset contains several housing and demographic features that may be used as input variables:

* `longitude`,`latitude`— geographic location of the district
* `housing_median_age`— age profile of housing stock
* `total_rooms`,`total_bedrooms`— housing supply within the district
* `population`,`households`— demographic density
* `median_income`— economic status of residents
* `ocean_proximity`— categorical location context (e.g., proximity to the coast)

Together, these features capture location, housing supply, demographic density, and economic conditions — all factors expected to influence district-level home values, both in 1990 and in principle today.


**5. Expected Outcome**

The expected outcome is a regression model that predicts `median_house_value` accurately on this historical dataset and generalizes well to unseen data drawn from the same distribution. The project's broader outcome is a validated, reusable ML pipeline — not a production-ready pricing tool for current markets.

**6. Evaluation Criteria**

**Technical metrics:**

* **Mean Absolute Error (MAE)**
* **Mean Squared Error (MSE)**
* **Root Mean Squared Error (RMSE)**

**Success criterion (within the scope of this dataset):**

A prediction error within roughly 10–15% of the actual median house value will be considered a strong result for this historical dataset, indicating the workflow is sound and worth extending to current data in the future.

**7. Assumptions & Constraints**

* **The dataset reflects a single historical snapshot (1990 California census) and does not represent current market prices or conditions.**
* **The data is aggregated at the district level, not the individual property level, so predictions represent district-wide averages rather than prices for specific homes.**
* **Important real-world pricing factors — such as exact property condition, school quality, crime rates, or recent comparable sales — are not present in this dataset and are out of scope for this project.**
* **Any future application of this model (or its methodology) to a real pricing platform would require retraining on current, live housing data.**

**8. Project Scope**

This project focuses on understanding the dataset, preparing the data for machine learning, exploring relationships between variables, building a suitable regression model, and evaluating its predictive performance — establishing a methodology intended for future reuse on current housing data.
